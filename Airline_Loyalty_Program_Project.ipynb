#Problem Statement
Northern Lights Air launched a loyalty-program promotion from February through April 2018 to increase customer enrollment. The goal of this analysis is to evaluate enrollment during the promotion and determine whether customers who joined during this period demonstrated different levels of retention, flight engagement, points activity, and customer lifetime value (CLV) compared with relevant historical customers.

The analysis also examines customer segments to identify patterns associated with higher customer value.

#Background
Northern Lights Air (NLA) is a fictional Canadian airline with a customer loyalty program that tracks enrollment, cancellations, flight activity, loyalty points, and customer demographics. From February through April 2018, NLA conducted a promotional campaign designed to increase loyalty-program enrollment.

#Business Objective
Northern Lights Air conducted a loyalty program promotion from February through April 2018. The objective of this analysis is to evaluate the effectiveness of the promotion and better understand the customers participating in the loyalty program.

The analysis will focus on answering the following business questions:
* Did loyalty-program enrollment increase during the promotion compared with previous years?
* How well were customers acquired during the promotion retained?
* How engaged were promotion customers through flight activity and loyalty-point activity?
* How does the value of promotion customers compare with a relevant historical customer group?
* Which customer segments are associated with higher customer lifetime value?

The goal is to provide NLA with insights that can help inform future loyalty promotions, customer retention efforts, and customer targeting strategies.

###How did enrollment during Feb–Apr 2018 compare with the same months in previous years?

```python
query = """
WITH monthly_enrollment_counts AS (
  SELECT
    enrollment_month,
    enrollment_year,
    COUNT(*) AS number_of_enrollments
  FROM loyalty_history
  WHERE enrollment_month IN (2,3,4)
  GROUP BY enrollment_month, enrollment_year),

historical_average AS (
  SELECT
    enrollment_month,
    AVG(number_of_enrollments) AS avg_enrollments
  FROM monthly_enrollment_counts
  WHERE enrollment_year BETWEEN 2013 AND 2017
  GROUP BY enrollment_month)

SELECT
  historical_average.enrollment_month,
  avg_enrollments,
  monthly_enrollment_counts.number_of_enrollments AS promotion_enrollments
FROM historical_average
LEFT JOIN monthly_enrollment_counts
  ON historical_average.enrollment_month = monthly_enrollment_counts.enrollment_month
WHERE monthly_enrollment_counts.enrollment_year == 2018
"""
pd.read_sql_query(query, conn)
```

###By what percentage did enrollment increase during each promotion month compared with its historical average?

```python
query = """
WITH monthly_enrollment_counts AS (
  SELECT
    enrollment_month,
    enrollment_year,
    COUNT(*) AS number_of_enrollments
  FROM loyalty_history
  WHERE enrollment_month IN (2,3,4)
  GROUP BY enrollment_month, enrollment_year),

historical_average AS (
  SELECT
    enrollment_month,
    AVG(number_of_enrollments) AS avg_enrollments
  FROM monthly_enrollment_counts
  WHERE enrollment_year BETWEEN 2013 AND 2017
  GROUP BY enrollment_month)

SELECT
  historical_average.enrollment_month,
  avg_enrollments,
  monthly_enrollment_counts.number_of_enrollments AS promotion_enrollments,
  ROUND(((monthly_enrollment_counts.number_of_enrollments - avg_enrollments)/avg_enrollments) * 100, 1) AS percent_increase
FROM historical_average
LEFT JOIN monthly_enrollment_counts
  ON historical_average.enrollment_month = monthly_enrollment_counts.enrollment_month
WHERE monthly_enrollment_counts.enrollment_year == 2018
"""
pd.read_sql_query(query, conn)
```

###How many customers in each enrollment type have a recorded cancellation?

```python
query = """
SELECT
  enrollment_type,
  COUNT(*) AS total_customers,
  COUNT(cancellation_year) AS cancelled_customers
FROM
  loyalty_history
GROUP BY enrollment_type
"""
pd.read_sql_query(query,conn)
```

###What percentage of customers in each enrollment type cancelled?

```python
query = """
SELECT
  enrollment_type,
  COUNT(*) AS total_customers,
  COUNT(cancellation_year) AS cancelled_customers,
  ROUND((COUNT(cancellation_year) * 100.0/COUNT(*)), 1) AS cancellation_rate
FROM
  loyalty_history
GROUP BY enrollment_type
"""
pd.read_sql_query(query,conn)
```

###Of the 971 customers acquired through the promotion, how many remained enrolled through the end of the dataset?

```python
query = """
SELECT
  enrollment_type,
  COUNT(*) AS total_customers,
  COUNT(cancellation_year) AS cancelled_customers,
  COUNT(*) - COUNT(cancellation_year) AS customers_retained,
  ROUND(((COUNT(*) - COUNT(cancellation_year))* 100.0/COUNT(*)), 1) AS observed_retention_rate
FROM
  loyalty_history
WHERE
  enrollment_type = '2018 Promotion'
GROUP BY enrollment_type
"""
pd.read_sql_query(query,conn)
```

###Did customers acquired through the 2018 Promotion fly more or less than other customers?

```python
query = """
SELECT
  loyalty_history.enrollment_type,
  COUNT(DISTINCT loyalty_history.loyalty_number) AS total_unique_customers,
  SUM(flight_activity.total_flights) AS flight_total,
  ROUND(SUM(flight_activity.total_flights) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number), 1) AS avg_flights_per_customer
FROM loyalty_history
LEFT JOIN flight_activity
  ON loyalty_history.loyalty_number = flight_activity.loyalty_number
WHERE
  loyalty_history.enrollment_type = '2018 Promotion' AND flight_activity.year = 2018
  AND flight_activity.month >= loyalty_history.enrollment_month
GROUP BY loyalty_history.enrollment_type
"""
pd.read_sql_query(query,conn)
```

###How did the flight activity of the 2018 Promotion cohort compare with customers who enrolled during Feb–Apr in previous years?

```python
query = """
SELECT
  enrollment_year,
  COUNT(DISTINCT loyalty_history.loyalty_number) AS total_unique_customers,
  SUM(flight_activity.total_flights) AS flight_total,
  ROUND(SUM(flight_activity.total_flights) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number), 1) AS avg_flights_per_customer
FROM
  loyalty_history
LEFT JOIN flight_activity
  ON loyalty_history.loyalty_number = flight_activity.loyalty_number
WHERE
  enrollment_type = '2018 Promotion'
  AND (
    (flight_activity.year - loyalty_history.enrollment_year) * 12
    + (flight_activity.month - loyalty_history.enrollment_month)
) BETWEEN 0 AND 8
GROUP BY enrollment_year
"""
pd.read_sql_query (query,conn)
```

```python
query = """
WITH clean_flight_activity AS (
  SELECT DISTINCT *
  FROM
    flight_activity
  ),

cohort_analysis AS (
  SELECT
    enrollment_year,
    COUNT(DISTINCT loyalty_history.loyalty_number) AS total_unique_customers,
    SUM(clean_flight_activity.total_flights) AS flight_total,
    ROUND(SUM(clean_flight_activity.total_flights) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number), 1) AS avg_flights_per_customer
  FROM
    loyalty_history
  LEFT JOIN clean_flight_activity
    ON loyalty_history.loyalty_number = clean_flight_activity.loyalty_number
  WHERE
    (
      (loyalty_history.enrollment_year = 2017 AND loyalty_history.enrollment_month IN (2,3,4)) OR loyalty_history.enrollment_type = '2018 Promotion'
      )
    AND (
      (clean_flight_activity.year - loyalty_history.enrollment_year) * 12
      + (clean_flight_activity.month - loyalty_history.enrollment_month)
  ) BETWEEN 0 AND 8
  GROUP BY enrollment_year)
SELECT
 (SELECT avg_flights_per_customer
 FROM cohort_analysis
 WHERE enrollment_year = 2018) AS twenty_eighteen_average,

 (SELECT avg_flights_per_customer
  FROM cohort_analysis
  WHERE enrollment_year = 2017) AS twenty_seventeen_average,

 Round((((SELECT avg_flights_per_customer
 FROM cohort_analysis
 WHERE enrollment_year = 2018)) - (SELECT avg_flights_per_customer
  FROM cohort_analysis
  WHERE enrollment_year = 2017)) * 100.0/(SELECT avg_flights_per_customer
  FROM cohort_analysis
  WHERE enrollment_year = 2017), 1) AS cohort_comparison
FROM
  cohort_analysis

"""
pd.read_sql_query(query,conn)
```

###Did customers acquired through the 2018 Promotion have different average CLV than a comparable prior-year cohort?

```python
query = """
SELECT
  enrollment_year,
  COUNT(DISTINCT loyalty_number) AS total_unique_customers,
  ROUND(AVG(clv),2) AS avg_clv
FROM
  loyalty_history
WHERE
  (enrollment_year = 2017 AND enrollment_month IN (2,3,4)) OR enrollment_type = '2018 Promotion'
GROUP BY
  enrollment_year
"""
pd.read_sql_query(query,conn)
```

```python
query = """
WITH clv_analysis AS (
  SELECT
  enrollment_year,
  COUNT(DISTINCT loyalty_number) AS total_unique_customers,
  ROUND(AVG(clv),2) AS avg_clv
FROM
  loyalty_history
WHERE
  (enrollment_year = 2017 AND enrollment_month IN (2,3,4)) OR enrollment_type = '2018 Promotion'
GROUP BY
  enrollment_year)

SELECT
  (SELECT avg_clv
  FROM clv_analysis
  WHERE enrollment_year = 2018) AS clv_twenty_eighteen_avg,

  (SELECT avg_clv
  FROM clv_analysis
  WHERE enrollment_year = 2017) AS clv_twenty_seventeen_avg,

  ROUND (((SELECT avg_clv
  FROM clv_analysis
  WHERE enrollment_year = 2018) - (SELECT avg_clv
  FROM clv_analysis
  WHERE enrollment_year = 2017))*100.0/(SELECT avg_clv
  FROM clv_analysis
  WHERE enrollment_year = 2017),1) AS percent_increase

"""

pd.read_sql_query(query,conn)
```

###Did the 2018 promotion cohort show greater loyalty-program activity than the comparable 2017 cohort?

```python
query = """
WITH clean_flight_activity AS (
    SELECT DISTINCT *
    FROM flight_activity
),

points_analysis AS (
  SELECT
    loyalty_history.enrollment_year,
    COUNT(DISTINCT loyalty_history.loyalty_number) AS total_unique_customers,
    SUM(clean_flight_activity.points_accumulated) AS total_points_accumulated,
    ROUND(SUM(clean_flight_activity.points_accumulated) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number)) AS avg_points_per_customer,
    ROUND(SUM(clean_flight_activity.points_redeemed) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number)) AS avg_points_redeemed_per_customer
  FROM loyalty_history
  LEFT JOIN clean_flight_activity
    ON loyalty_history.loyalty_number = clean_flight_activity.loyalty_number

  WHERE
    (
      (loyalty_history.enrollment_year = 2017 AND loyalty_history.enrollment_month IN (2,3,4)) OR loyalty_history.enrollment_type = '2018 Promotion'
      )
    AND (
      (clean_flight_activity.year - loyalty_history.enrollment_year) * 12
      + (clean_flight_activity.month - loyalty_history.enrollment_month)
  ) BETWEEN 0 AND 8

  GROUP BY enrollment_year)

SELECT *
FROM points_analysis

"""

pd.read_sql_query(query,conn)




```

```python
query = """
WITH clean_flight_activity AS (
    SELECT DISTINCT *
    FROM flight_activity
),

points_analysis AS (
  SELECT
    loyalty_history.enrollment_year,
    COUNT(DISTINCT loyalty_history.loyalty_number) AS total_unique_customers,
    SUM(clean_flight_activity.points_accumulated) AS total_points_accumulated,
    ROUND(SUM(clean_flight_activity.points_accumulated) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number)) AS avg_points_per_customer,
    ROUND(SUM(clean_flight_activity.points_redeemed) * 1.0/COUNT(DISTINCT loyalty_history.loyalty_number)) AS avg_points_redeemed_per_customer
  FROM loyalty_history
  LEFT JOIN clean_flight_activity
    ON loyalty_history.loyalty_number = clean_flight_activity.loyalty_number

  WHERE
    (
      (loyalty_history.enrollment_year = 2017 AND loyalty_history.enrollment_month IN (2,3,4)) OR loyalty_history.enrollment_type = '2018 Promotion'
      )
    AND (
      (clean_flight_activity.year - loyalty_history.enrollment_year) * 12
      + (clean_flight_activity.month - loyalty_history.enrollment_month)
  ) BETWEEN 0 AND 8

  GROUP BY enrollment_year)

SELECT
  MAX(CASE WHEN enrollment_year = 2018
         THEN avg_points_per_customer
    END) AS twenty_eighteen_avg_points,
  MAX(CASE WHEN enrollment_year = 2017
         THEN avg_points_per_customer
    END) AS twenty_seventeen_avg_points,
  ROUND((MAX(CASE WHEN enrollment_year = 2018
         THEN avg_points_per_customer
    END) - MAX(CASE WHEN enrollment_year = 2017
         THEN avg_points_per_customer
    END)) * 100.0/MAX(CASE WHEN enrollment_year = 2017
         THEN avg_points_per_customer
    END),1) AS percent_increase
FROM points_analysis

"""

pd.read_sql_query(query,conn)
```
